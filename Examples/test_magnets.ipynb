```python
import parastell.magnet_coils as magnet_coils
import cubit
import numpy as np
```

```python
coils_file = 'coils_wistelld.txt'
circ_cross_section = ['circle', 20]
toroidal_extent = 90.0
sample_mod = 6

circ_coil_obj = magnet_coils.MagnetSet(
    coils_file,
    circ_cross_section,
    toroidal_extent,
    sample_mod = 6
)
```

```python
circ_coil_obj.build_coil_surface()
```

```python
circ_coil_obj.build_magnet_coils()
```

```python
cubit.cmd('save cub5 "coiltest.cub5" overwrite')
```

```python
circ_coil_obj.export_step()
```

```python
circ_coil_obj.filtered_filaments.shape
```

```python
filaments = circ_coil_obj.filtered_filaments
```

```python
test = filaments
filaments[:,0,:].shape
```

```python
flattened = []
for i, filament in enumerate(test):
    sublist = []
    for k, point in enumerate(filament):
        sublist.append(str(point))
    flattened.append(sublist)

flattened = np.array(flattened)
```

```python
import pandas as pd
df = pd.DataFrame(flattened.T)
df.to_csv('filaments.csv')
```
